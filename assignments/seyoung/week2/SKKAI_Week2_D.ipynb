```python

import os
import shutil

src = "/content/work/qwen1.5b-q3_k_m.gguf"
dst_dir = "/content/drive/MyDrive/skkai_w2/gguf"
dst = os.path.join(dst_dir, "qwen1.5b-q3_k_m.gguf")

print("원본 파일 존재:", os.path.isfile(src))
print("공유 폴더 존재:", os.path.isdir(dst_dir))

if os.path.isfile(src) and os.path.isdir(dst_dir):
    print("파일 복사 시작")
    shutil.copy2(src, dst)

    print("업로드 파일 존재:", os.path.isfile(dst))
    print("크기 일치:", os.path.getsize(src) == os.path.getsize(dst))
    print("파일 크기(GB):", round(os.path.getsize(dst)/1e9, 4))
else:
    print("원본 파일 또는 공유 폴더를 찾지 못했습니다.")

```
